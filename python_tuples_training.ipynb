## 1. Introduction to Tuples
- Tuples are immutable sequences in Python.
- They are defined using parentheses `()`.
- Unlike lists, tuples cannot be modified after creation.

```python
# Creating tuples
empty_tuple = ()
single_tuple = (5,)
multi_tuple = (1, 2, 3, 4)

print("Empty:", empty_tuple)
print("Single:", single_tuple)
print("Multiple:", multi_tuple)
```

## 2. Accessing Elements
- Indexing starts at 0.
- Negative indexing starts from the end.
- Slicing extracts parts of a tuple.

```python
sample_tuple = (10, 20, 30, 40, 50)
print("First element:", sample_tuple[0])
print("Last element:", sample_tuple[-1])
print("Slice (1:4):", sample_tuple[1:4])
```

## 3. Packing and Unpacking
- Packing: grouping values into a tuple.
- Unpacking: extracting values from a tuple.

```python
# Packing
packed = ("Alice", 25, "Engineer")
print("Packed tuple:", packed)

# Unpacking
name, age, profession = packed
print("Name:", name)
print("Age:", age)
print("Profession:", profession)
```

## 4. Tuple Methods
- `count()` → counts occurrences
- `index()` → finds the position of an element

```python
numbers = (1, 2, 3, 2, 4, 2)
print("Count of 2:", numbers.count(2))
print("Index of 3:", numbers.index(3))
```

## 5. Tuple Operations
- Concatenation: `+`
- Repetition: `*`
- Membership: `in`, `not in`

```python
tuple1 = (1, 2)
tuple2 = (3, 4)
print("Concatenation:", tuple1 + tuple2) # you can also use 
print("Repetition:", tuple1 * 3)
print("Membership (2 in tuple1):", 2 in tuple1)
```
